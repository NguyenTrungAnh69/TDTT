### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
def new_function():
    n = int(input())
    count = 0
    for k in range(1,n):
        if k > ( n - 1 ) / 2:
            count += 1
    return count

new_function()
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
def triangle_area():
    a, b, c = map(float, input().split())
    if a + b > c and b + c > a and c + a > b:
        p = (a + b + c) / 2
        area = (p * (p - a) * (p - b) * (p - c)) ** 0.5
        return area
    else: raise ValueError("The sides cannot form a triangle")

triangle_area()
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
def shift_char():
  while True:
    x = input("Base character: ")
    if len(x) == 1 and x.isalpha():
      break
    print("1st input must be a character")
  n = int(input("How many positions after base character: "))
  if x.isupper():
    base = ord("A")
  else:
    base = ord("a")
  new_pos = (ord(x) - base + n) % 26
  return chr(base + new_pos)

shift_char()
```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
def convert_to_lower():
    x = input("Single character:")
    if len(x) == 1 and x.isalpha():
        # x.isalpha = not x.isdigit
        return x.lower()
    else:
        raise ValueError("Input must be a single character")

convert_to_lower()
```

### Problem 10
#### Write a function that, given a lowercase letter, returns a string containing that letter repeated 10 times.

```python
def letter_repeat():
    while True:
        x = input ("Single character:")
        if len(x) == 1 and x.islower():
            break
        print ("Input must be a lowercase letter")
    return x * 10

letter_repeat()
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
def right_star_triangle():
    n = int (input())
    star_list = []
    for i in range(1, n + 1):
        star_list.append('*' * i)
    print ("\n".join(star_list))

right_star_triangle()
```
