### Problem 2

#### Một khu đất hình chữ nhật có chiều rộng a, chiều dài b. Người ra xây một khu vui chơi hình tròn tại vị trí trung tâm khu đất như hình vẽ. Nhập vào a, b. Hãy tính phần diện tích còn để trồng cây xung quanh là bao nhiêu. Biết pi = 3.14, lấy kết quả đến 2 chữ số phần thập phân.

```python
a = float(input("Rectangle_Width"))
b = float(input("Rectangle_Length"))

Rectangle_Area = a * b
Circle_Area = 3.14 * ((a/2)**2)
Tree_planting_Area = Rectangle_Area - Circle_Area

print (f"{Tree_planting_Area:.2f}")
```

### Problem 3

#### Viết chương trình nhập vào một “kí tự” c, nếu nó là chữ hoa thì in ra chữ thường tương ứng và ngược lại nếu nó là chữ thường thì in ra chữ hoa tương ứng của nó. (xem trước cấu trúc lựa chọn: if….else, ma)

```python
def convert_c(x):
    while True:
        if x == 'c' or x == 'C':
            break
        raise ValueError("Input must be 'c' or 'C'")
    if x.isalpha() and x.islower():
        return x.upper()
    else:
        if x.isalpha() and x.isupper():
            return x.lower()
    return x

char = input()
convert_c(char)
```

### Problem 4

#### Viết chương trình nhập vào một “kí tự”, Kiểm tra xem c có phải là kí tự chữ cái không (hoa hoặc thường). Nếu đúng thì in ra “{c} là kí tu alphabet”, nếu sai in ra “{c} không phải là kí tự alphabet.

```python
def alphabet_check(x):
    while True:
        if len(x) == 1:
            break
        raise ValueError ("Input là một kí tự")
    if x.isalpha():
        return f"{x} là kí tự alphabet"
    else:
        return f"{x} không phải là kí tự alphabet"

char = input()
alphabet_check(char)
```

### Problem 5

#### Bài 5: Nhập vào chữ cái hoa A, in ra chữ cái thường liền trước chữ cái a thường tương ứng của A. Xét TH khi c = ‘A’ là đặc biệt.

```python
def new_function():
    while True:
        x = input()
        if x.isupper() and len(x) == 1:
            break
        print("Input must be an upper character")
    if x != 'A':
        return chr((ord(x)-1)).lower()
    else: 
        return "Đặc biệt"

new_function()
```

### Problem 6

#### Nhập vào 3 số a, b, c. Kiểm tra xem đây có phải là cạnh của 1 tam giác không, nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó không phải 3 cạnh của tam giác thì in ra “Khong phai 3 canh cua tam giac”.

```python
def triangle_check():
    while True:
        a, b, c = map(float, input().split())
        if a + b > c and a + c > b and b + c > a:
            break
        print ("Khong phai 3 canh cua tam giac")
    p = (a + b + c) / 2
    area = (p * (p - a) * (p - b) * (p - c)) ** 0.5
    return f"{area:.1f}"

triangle_check()
```
